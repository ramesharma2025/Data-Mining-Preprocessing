```python
import pandas as pd

df = pd.read_csv('student-record.csv')

# Clean: Convert column to numeric (handling any spaces/noisy artifacts)
df['Total_Marks'] = pd.to_numeric(df['Total_Marks'], errors='coerce')

# Standardize: Subtract the mean and divide by the standard deviation
mean_val = df['Total_Marks'].mean()
std_val = df['Total_Marks'].std()

df['Total_Marks_Zscore'] = (df['Total_Marks'] - mean_val) / std_val

print(df[['Total_Marks', 'Total_Marks_Zscore']].head(50))
```

```python
from sklearn.preprocessing import StandardScaler

df_scaled = df.copy()
num_cols = ['Assessment_Score', 'Total_Marks']

# Coerce columns to numeric format
df_scaled[num_cols] = df_scaled[num_cols].apply(pd.to_numeric, errors='coerce')

# Apply Z-score scaling
df_scaled[num_cols] = StandardScaler().fit_transform(df_scaled[num_cols])

print(df_scaled[num_cols].head())
```
