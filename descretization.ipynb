```python
import pandas as pd

df = pd.read_csv('student-record.csv')
df['Total_Marks'] = pd.to_numeric(df['Total_Marks'], errors='coerce')

# Discretize continuous 'Total_Marks' into conceptual labels
bins = [0, 60, 75, 90, 500]
labels = ['Needs Improvement', 'Average', 'Good', 'Excellent']

df['Performance_Category'] = pd.cut(df['Total_Marks'], bins=bins, labels=labels)
print(df[['Total_Marks', 'Performance_Category']].head())
```
