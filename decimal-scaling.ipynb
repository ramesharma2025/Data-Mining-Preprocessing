```python
import pandas as pd
import numpy as np

df = pd.read_csv('student-record.csv')

# Clean: Convert column to numeric (handling any hidden string noise/spaces)
df['Total_Marks'] = pd.to_numeric(df['Total_Marks'], errors='coerce')

# Find 'j': Count digits of the maximum absolute value in the column
max_val = df['Total_Marks'].abs().max()
j = len(str(int(max_val)))  # Finds how many digits the maximum value has

# Apply Decimal Scaling: Divide by 10^j
df['Total_Marks_Scaled'] = df['Total_Marks'] / (10**j)

print(f"Chosen j value (multiplier power): {j}")
print(df[['Total_Marks', 'Total_Marks_Scaled']].head())
```
